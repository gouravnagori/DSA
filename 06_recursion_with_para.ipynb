```python
# Recursion using parameter
```

```python
# you have to print x(any integer) n times by resursion

def fun(x , n):
    if n == 0:  #this is the base condition
        return
    print(x)
    fun(x,n-1)
fun(69,5)    
```

```python
# you have to print 1 to n using resursion
# this is tail recursion (backtracking)
def func(n):
    if n == 0:
        return
    func(n-1)
    print(n)
func(6)    
```

```python
# you have to print i to n using resursion
# and this is head recursion
def func(i,n):
    if n < i:
        return
    print(i)
    func(i+1 , n)
func(2,6)  
```
