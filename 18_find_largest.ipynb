```python
# Find largest element in an array
```

```python
arr = [55,23,-22,-32,11,0,99,32,75]
```

```python
# this is the built in method 
max(arr)
```

```python
# Method -1 
largest = arr[0]
for i in arr:
    if i>largest:
        largest = i
largest        
```

```python
arr2 = [-18,-73,-9,-21]
largest = arr2[0]
for i in arr2:
    if i>largest:
        largest = i
largest        
```

```python
# Another approch for Method -1 
largest = arr[0]
for i in arr:
    largest = max(largest,i)
largest
```

```python
TC : O(N)
SC : O(1)
```

```python
# m-2 in this we take the value of largest as - infinite 

arr3 = [23,0,78,-18,-73,-9,-21]
largest = float("-inf")
for i in arr3:
    if i>largest:
        largest = i
largest        
```

```python
TC : O(N)
SC : O(1)
```
