```python
arr3 = [12,32,23,2,45,7,88,5,333,333]
largest = float("-inf")
S_largest = float("-inf")
for i in arr3:
    if i>largest:
        S_largest = largest
        largest = i
    elif i>S_largest and i != largest:
        S_largest = i 
S_largest        
```
