## Extraction of digits via loop

```python
# Extraction of digits via loop

n = int(input("Enter a Number: "))
num = n
while num > 0:
    digit = num%10
    print(digit, end =" ")
    num = num//10
```

```python
# Count number of digit

n = int(input("Enter a Number: "))
num = n
count = 0
while num > 0:
    count += 1
    num = num//10
print(f"\n{n} has {count} number of digit.")    
```

```python
import math

num = int(input("Enter a number: "))

if num > 0:
    digits = int(math.log10(num)) + 1
    print("Number of digits =", digits)
else:
    print("Please enter a positive number.")
```

```python
# check the number is palindrome or not
n = int(input("Enter the number: "))
num = n
result = 0
while num > 0:
    ld = num%10
    result = (result*10) + ld
    num = num//10
if result == n:
    print("Enterd Number is Palindrome")
```
