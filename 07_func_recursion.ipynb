```python
# Parameterized and function recursion
```

```python
# sum of 1 to n

def func(sum,i,n):
    if i>n:
        print(sum)
        return
    func(sum+i,i+1,n)
func(0,1,4)    
```

```python
# function recursion
#sum of 1 to n

def func(n):
    if n == 1:
        return 1
    return n + func(n-1)

func(10)    
```

```python
# factorial program

def func(n):
    if n == 1 or n == 0:
        return 1
    return n * func(n-1)

func(6)  
```
