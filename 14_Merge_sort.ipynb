# Merge Sort

```python
# Program to merge two sorted array

def merge_arr(left , right):
    result = []
    i,j = 0,0
    n,m = len(left) , len(right)
    while i<n and j<m:
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j +=1

    while i < n:
        result.append(left[i])
        i += 1

    while j < m:
        result.append(right[j])
        j += 1

    return result

left = [1,2,4,5,8,9]   
right = [3,4,5,8,9,9,10,11]

merge_arr(left,right)
```

```python
def merge_arr(left , right):
    result = []
    i,j = 0,0
    n,m = len(left) , len(right)
    while i<n and j<m:
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j +=1

    while i < n:
        result.append(left[i])
        i += 1

    while j < m:
        result.append(right[j])
        j += 1

    return result


def MergeSort(nums):
    if len(nums) <= 1:
        return nums
    mid = len(nums) // 2
    left_arr = nums[:mid]
    right_arr = nums[mid:]
    left = MergeSort(left_arr)
    right = MergeSort(right_arr)
    return merge_arr(left , right)

MergeSort([2,12,2,40,5,7,4,2,3,7,8,5,3,2])    
```
