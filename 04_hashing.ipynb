```python
# method - 1 : Brute force

n = [5,3,2,2,1,5,5,7,5,10]
m = [10,11,1,9,5,67,2]

for i in m:
    count = 0
    for x in n:
        if x == i :
            count +=1
    print(count)            
```

```python
# method - 2
n = [5,3,2,2,1,5,5,7,5,10]
m = [10,11,1,9,5,67,2]

hash_list = [0] * 11
for num in n:
    hash_list[num] += 1 

for num in m:
    if num<1 or num>10:
        print(0)
    else:
        print(hash_list[num])
```

```python
# method - 3
# we can do the same question with the help of dict also

n = [5,3,2,2,1,5,5,7,5,10]
m = [10,11,1,9,5,67,2]
hash_dict = {}
num = len(n)
for i in range(0,num):
    hash_dict[n[i]] = hash_dict.get(n[i] , 0) + 1 
hash_dict    
```

```python
for i in m:
    if i in n:
        print(hash_dict[i])
    else:
        print(0)
```


## character hashing



```python
s = "azyxyyidfyhiuofsdsssssssssssss"
q = ["d","a","y","l","s","i"]
# same question as above
```

```python
# Method - 1 (brute force method)
s = "azyxyyidfyhiuofsdsssssssssssss"
q = ["d","a","y","l","s","i"]
for i in q:
    count = 0
    for n in s:
        if n == i:
            count += 1
    print(count)    
```

```python
# so for m-1 
# TC is O(m*n)
# TLE error
```

```python
s = "azyxyyidfyhiuofsdsssssssssssss"
q = ["d","a","y","l","s","i"]

hash_list = [0]*26

for ch in s:
    ascii_val = ord(ch)
    index = ascii_val - 97
    hash_list[index] += 1

for ch in q:
    ascii_val = ord(ch)
    index = ascii_val - 97
    print(hash_list[index])
```
