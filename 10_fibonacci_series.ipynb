```python
# Fibonacci series via recursion
```

```python
def fibo(n):
    if n <= 0:
        return n 
    if n == 1:
        return n
    return fibo(n-1) + fibo(n-2)   
fibo(6)    
```

```python
# TC O(2^n)
# ans SC is O(N)
```

```python
# leetcode 509
class Solution:
    def func(self, num):
        if num == 0 or num == 1:
            return num
        return self.func(num - 1) + self.func(num - 2)

    def fib(self, n):
        return self.func(n)


solution = Solution()
solution.fib(6)
```
