```python
# Head resursion

count = 0
def greet():
    global count
    if count == 5:
        return
    else:
        count += 1
        print("Sweet Dream")
        greet()   

greet()    
```

```python
# Tail resursion
count = 0
def greet():
    global count
    if count == 5:
        return
    count +=1
    greet()
    print("Jay Shree Krishna")

greet()
```
