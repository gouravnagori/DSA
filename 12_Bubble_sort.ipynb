# Bubble Sort

```python
def Bubble_sort(nums):
    n = len(nums)
    for i in range(n-2,-1,-1):
        for j in range(0,i+1):
            if nums[j] > nums[j+1]:
                nums[j],nums[j+1] = nums[j+1],nums[j]
    return nums
nums = [42,2,54,63,3,2,8]
Bubble_sort(nums)
```

```python
def Bubble_sort(nums):
    n = len(nums)
    for i in range(n-2,-1,-1):
        is_swap = False
        for j in range(0,i+1):
            if nums[j] > nums[j+1]:
                nums[j],nums[j+1] = nums[j+1],nums[j]
                is_swap = True
        if is_swap == False:
            break
    return nums
nums = [1,2,3,4,5,6,7,8]
Bubble_sort(nums)
```

```python
# Another approch of writing same code 
# we can aslo write range like this for better understanding

def Bubble_sort(nums):
    n = len(nums)
    for i in range(0,n-1):
        is_swap = False
        for j in range(0,n-i-1):
            if nums[j] > nums[j+1]:
                nums[j],nums[j+1] = nums[j+1],nums[j]
                is_swap = True
        if is_swap == False:
            break
    return nums
nums = [1,2,3,32,6,7,6]
Bubble_sort(nums)
```
