```python
def update(lst):
    n = len(lst)
    i = 0  
    
    while i < n: 
        if lst[i] == 0:
            for j in range(i, n-1):
                lst[j+1], lst[j] = lst[j], lst[j+1]
            n -= 1 
        else:
            i += 1 
            
    return lst
update([0,3,0,4,2,0,0,9,7])
```

```python
def update(lst):
    insert_pos = 0
    
    # Step 1: Move all non-zero elements to the front
    for num in lst:
        if num != 0:
            lst[insert_pos] = num
            insert_pos += 1
            
    # Step 2: Fill the rest of the list with zeros
    while insert_pos < len(lst):
        lst[insert_pos] = 0
        insert_pos += 1
        
    return lst

# Test case
print(update([0, 3, 0, 4, 2, 0, 0, 9, 7]))
# Output: [3, 4, 2, 9, 7, 0, 0, 0, 0]
```

```python
def update_pythonic(lst):
    # False (0) comes before True (1)
    lst.sort(key=lambda x: x == 0) 
    return lst

update_pythonic([0,3,0,4,2,0,0,9,7])
```

```python
def update(list):
    insert_pos = 0
    for i in range(len(list)):
        if list[i] != 0:
            list[insert_pos], list[i] = list[i], list[insert_pos]
            insert_pos += 1
    return list
update([0,3,0,4,2,0,0,9,7])    
```
