```python
# Check the string is palindrome or not via recursion and loop
```

```python
# Using loop

def func(s,left,right):
    while left < right:
        if s[left] != s[right]:
            return False
        left += 1
        right -= 1
    return True    
s = "abcddcba"
left = 0 
right = len(s) -1
func(s,left,right)
```

```python
# via loop
# TC = O(N/2) or we can say O(N)
# SC = O(1)
```

```python
# Using recursion

def func(s, l , r):
    if s[l] != s[r]:
        return False
    if l >= r :
        return True
    return func(s, l+1, r-1)

s = "asddsa"
func(s , 0 ,len(s)-1)    

    
```

```python
# for recursion
# TC = O(N/2) or we can say O(N)
# SC = O(N/2) or we can say O(N) that is stack space
```
