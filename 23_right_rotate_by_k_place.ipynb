```python
# OPTIMAL APPROCH

def rotate(self, nums, k):
    n = len(nums)
    k = k % n

    nums.reverse()
    nums[:k] = reversed(nums[:k])
    nums[k:] = reversed(nums[k:])
```

```python
def reverse(nums, left, right):
    while left<right:
        nums[left],nums[right] = nums[right],nums[left]
        left += 1
        right -= 1 
        
nums = [1,2,3,4,5,6]
n = len(nums)
k = 3                      # TC
reverse(nums , n-k , n-1)  # k/2
reverse(nums , 0,n-k-1)    # (n-k)/2
reverse(nums , 0 , n-1)    # n/2
nums
```
