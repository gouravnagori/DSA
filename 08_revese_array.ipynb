```python
def rev(arr,l,r):
    if l >= r:
        return arr  
    arr[l],arr[r] = arr[r],arr[l]
    return rev(arr , l+1 , r-1)

arr = [1,2,3,4,5,66,3]
rev(arr, 0 , (len(arr) -1))
```

```python
# GATE 2026 - QUE

def mystery(n):
    if n <= 0:
        return 1
    else:
        return mystery(n-1) + mystery(n-2)

mystery(4)
```
