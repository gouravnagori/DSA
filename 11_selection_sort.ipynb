# Selection Sort

```python
def selection_sort(nums):
    for i in range(0 , len(nums)):
        min_indx = i
        for j in range(i+1 , len(nums)):
            if nums[j] < nums[min_indx]:
                min_indx = j
        nums[i],nums[min_indx] = nums[min_indx],nums[i]
    return nums

nums = [2,4,1,3,2,8,7]
selection_sort(nums)    
```

```python
# TC = O(n(n+1))/2 = O(n^2)
# SC = O(1)
```

```python
# descending order (largest → smallest)

arr = [64, 25, 12, 22, 11]
n = len(arr)
for i in range(n):
    max_index = i

    for j in range(i + 1, n):
        if arr[j] > arr[max_index]:
            max_index = j

    arr[i], arr[max_index] = arr[max_index], arr[i]

arr
```
