```python
arr = [1,1,1,2,3,6,7,7,7,8,10,10]
l = len(arr)
i = 0

while i < len(arr) - 1:
    if arr[i] == arr[i + 1]:
        arr.pop(i + 1)
    else:
        i += 1
```

```python
arr = [1,1,1,2,3,6,7,7,7,8,10,10]

l = len(arr)

i = 0
j = 1

while j < l:
    if arr[i] != arr[j]:
        i += 1
        arr[i], arr[j] = arr[j], arr[i]
    j += 1

print(arr)
print(i + 1)
```
