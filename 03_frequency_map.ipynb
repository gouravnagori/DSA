```python
num = [1,2,3,2,2,34,21,3,211,2,2,23,34,45,23,1,4,1,1,2,2,2,23,23,23,2,32,2,89]
freq_dict = dict()
for i in range(0,len(num)):
    if num[i] not in freq_dict:
        freq_dict[num[i]] = 1
    else:
        freq_dict[num[i]] += 1
print(freq_dict)    
```

```python
# method - 2
# this is the easy to write the same code 
# and tc is same as above that is O(N)
num = [1,2,3,2,2,34,21,3,211,2,2,23,34,45,23,1,4,1,1,2,2,2,23,23,23,2,32,2,89]
hash_map = {}
n = len(num)
for i in range(0,n):
    hash_map[num[i]] = hash_map.get(num[i] , 0) + 1
hash_map
```
