```python
# code - 2
n = int(input("Enter the number: "))
num = n 
result = [i for i in range(1,num//2 + 1) if num%i == 0]
result.append(num)
print(result)
```

```python
# code 2 in more efficient than code 1 but not as much optimal 
# because the time complexity is almost same in the both case that
# is O(N)
# so we have to make it more efficient or optimal so we have to 
# write our code in more better way
```

```python
# code - 3
from math import sqrt
n = int(input("Enter the number: "))
result = []
num = n 
for i in range(1, int(sqrt(num))+1):
    if num%i == 0:
        result.append(i)
        if num//i != i:
            result.append(num//i)
result.sort()
result
```

```python
# for this code the tc is O(root N) + O(Nlog N)
# and if we don't include sort method then the tc is O(root N)
# and code - 3 is the best or i can say optimal code.
```
